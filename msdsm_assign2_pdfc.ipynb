```python
# 7.Write a Python function that takes a sequence of numbers and determines whether all the numbers are different from each other.
def test_dist(data):
    if len(data) == len(set(data)):
        return True
    else:
        return False
x = list(input("please enter the data with <space> in it: "))
print(f"is the data has different integer '{test_dist(x)}'")
```
